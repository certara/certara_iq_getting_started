```python
import pandas as pd
from plotnine import *
```

```python
Tpar_joint = pd.read_csv('inputs/Analyze_Optimization_completed.ipynb/Tpar_fit_joint.csv')
Tpar_ind = pd.read_csv('inputs/Analyze_Optimization_completed.ipynb/Par_fit_ind.csv')
```

```python
Tpar_joint2 = Tpar_joint.query('is_fit == True')
Tpar_ind2 = Tpar_ind.query('is_fit == True')
```

```python
fig = (ggplot()
       + geom_point(data = Tpar_ind2, mapping = aes(x = 'parameter', y = 'value'), position = position_jitter(width = 0.15, height = 0))
       + geom_point(data = Tpar_joint2, mapping = aes(x = 'parameter', y = 'value'), shape = '*',
                    color = 'black', stroke = 0.2, fill = 'goldenrod', size = 6)
       + labs(x = 'Parameter Name', y = 'Parameter Value')
      )

display(fig)
```
